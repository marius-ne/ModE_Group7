# Minimalbeispiel mit einem Plot und anschließendem Speichern

```python
lambda_in_min = 0.173
lambda_out_min = 0.2
Q_out_nom = 530
eta_nom = 0.8

Q_in_nom = Q_out_nom / eta_nom

Q_out_min = lambda_out_min * Q_out_nom
Q_in_min = lambda_in_min * Q_out_nom / eta_nom

eta_min = lambda_out_min * eta_nom / lambda_in_min

eta_mean = (eta_nom + eta_min) / 2


apply_style(grid=False, latex=False)

Q_in = np.linspace(0, Q_in_nom, 200)

fig, ax = plt.subplots()

ax.plot(Q_in, eta_nom * Q_in, label=rf"$\eta$ = {eta_nom:.2f}")
ax.plot(Q_in, eta_min * Q_in, label=rf"$\eta$ = {eta_min:.2f}")
ax.plot(Q_in, eta_mean * Q_in, label=rf"$\eta$ = {eta_mean:.2f}")

ax.axhline(y=Q_out_nom, color="red", linestyle="--")

ax.scatter(
    [Q_in_nom, Q_in_min],
    [Q_out_nom, Q_out_min],
    color="green",
    zorder=3,
)

ax.set_xlabel(r"$\dot{Q}_{\rm{in}}$")
ax.set_ylabel(r"$\dot{Q}_{\rm{out}}$")
ax.legend()
ax.grid(True)

plt.show()

safe_figure(fig, "Plots", "test")
```

# Minimalbeispiel mit zwei Subplots

```python
apply_style(ncols=2, grid=False, latex=False)

# Beispieldaten
x = np.linspace(0, 2 * np.pi, 200)
y1 = np.sin(x)
y2 = np.cos(x)

# Figure mit zwei Subplots
fig, (ax1, ax2) = plt.subplots(
    1, 2,
    sharey=True,
    constrained_layout=True,
)

# Linker Plot
ax1.plot(x, y1, label=r"$\sin(x)$")
ax1.set_xlabel(r"$x$")
ax1.set_ylabel(r"$y$")
ax1.legend()

# Rechter Plot
ax2.plot(x, y2, label=r"$\cos(x)$", color="tab:orange")
ax2.set_xlabel(r"$x$")
ax2.legend()

plt.show()
```
